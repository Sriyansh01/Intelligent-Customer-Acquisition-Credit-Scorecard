```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Main Dataset

pd.set_option('display.max_columns', None)
df = pd.read_csv('../data/raw/application_train.csv')
df.shape
```

```python
# Basic Structure Check

df.head()
```

```python
df.info()
```

```python
# Understand Target Variable

df['TARGET'].value_counts()
sns.countplot(x='TARGET', data=df)
plt.title("Target Distribution")
plt.show()
```

```python
# Missing Value Audit

missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

missing.head(20)

# calculate percentage:

missing_percent = (missing / len(df)) * 100
missing_percent.head(20)
```

```python
# Identify Categorical vs Numerical

categorical_cols = df.select_dtypes(include=['object']).columns
numerical_cols = df.select_dtypes(exclude=['object']).columns

print("Categorical:", len(categorical_cols))
print("Numerical:", len(numerical_cols))
```

```python
# Check Strange Values (Very Important)

df[['DAYS_BIRTH','DAYS_EMPLOYED']].describe()
```

```python
# Quick Correlation With Target

numerical_df = df.select_dtypes(include=[np.number])

correlations = numerical_df.corr()['TARGET'].sort_values()

correlations.head(10)
correlations.tail(10)
```
